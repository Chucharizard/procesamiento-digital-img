```python
import cv2 
import numpy as np
import matplotlib.pyplot as plt
img = cv2.imread('colores.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

fig, axs = plt.subplots(1, 4, figsize=(20, 5))

axs[0].imshow(img_rgb)
axs[0].set_title('Imagen Original')

fig_img=axs[1].imshow(img_rgb[...,0], cmap='gray')
fig.colorbar(fig_img, ax=axs[1], shrink=0.5)
axs[1].set_title('Canal Rojo')

fig_img=axs[2].imshow(img_rgb[...,1], cmap='gray')
fig.colorbar(fig_img, ax=axs[2], shrink=0.5)
axs[2].set_title('Canal Verde')

fig_img=axs[3].imshow(img_rgb[...,2], cmap='gray')
fig.colorbar(fig_img, ax=axs[3], shrink=0.5)
axs[3].set_title('Canal Azul')
```

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams['figure.figsize'] = (10, 5)
size_img = 100
r=np.ones((size_img, size_img))*255
g=np.ones((size_img, size_img))*128
b=np.ones((size_img, size_img))*0

imagen=np.dstack((r,g,b))
imagen=np.array(imagen, dtype=int)

plt.figure()
plt.imshow(imagen)
plt.title('imagen de color ')
plt.show
```

Escala de grises y profundidad de color

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('colores.jpg', cv2.IMREAD_GRAYSCALE)
# valores iniciales de profundidad del color 
original = cv2.imread('colores.jpg')

img16bits=np.uint16(img)
img_new=img16bits.dtype
print(f'La imagen original es de tipo: {img_new}')
#guardado con cv2
#guardado con pyplot 

cv2.imwrite('imagen.png', img) 
plt.imsave('imagen_16bits.png', cv2.cvtColor(img16bits, cv2.COLOR_BGR2RGB)) #guardar 
#imagen estable tratada 

plt.figure(figsize= (10,5))
plt.subplot(1,2,1)
plt.title('Imagen original')
plt.imshow(img, cmap='gray')

plt.subplot(1,2,2)
plt.title('Imagen 16 bits')
plt.imshow(img16bits, cmap='gray')
```
