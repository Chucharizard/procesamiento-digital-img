# Laboratorio 2 - Imagen RGB con 4 Cuadrantes
Creación de una imagen de **100×100 píxeles** dividida en 4 cuadrantes de colores usando `numpy` y ciclos `for`.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 100
mitad = N // 2   

print(f"Tamaño de todo  : {N}×{N}")
print(f"La  mitad         : {mitad}")
print(f"Cada cuadrante: {mitad}×{mitad}")
```

```python
imagen = np.zeros((N, N, 3), dtype=np.uint8)
print("Shape de la imagen:", imagen.shape)
```

```python
imagen[0:mitad, 0:mitad]  = [0,   255, 0  ]  
imagen[0:mitad, mitad:N]  = [0,   0,   255]  
imagen[mitad:N, 0:mitad]  = [255, 0,   0  ]  
imagen[mitad:N, mitad:N]  = [0,   255, 255]  

print("Imagen rellenada correctamente")
```

```python

plt.figure(figsize=(4, 4))
plt.imshow(imagen)
plt.title("Imagen 100×100 – 4 Cuadrantes RGB")
plt.axis("off")
plt.tight_layout()
plt.savefig("imagen_cuadrantes.png", dpi=100)
plt.show()
```
