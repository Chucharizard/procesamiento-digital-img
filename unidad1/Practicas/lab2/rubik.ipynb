# Laboratorio 2 - Cara del Cubo Rubik (9 cuadrantes)
Imagen de **99×99 píxeles** dividida en una grilla **3×3** usando slicing de NumPy.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N = 99
s = N // 3   

print(f"Tamaño total    : {N}×{N}")
print(f"Sección (s)     : {s}")
print(f"Cada cuadrante  : {s}×{s}")

# Canvas en negro
imagen = np.zeros((N, N, 3), dtype=np.uint8)
print("Shape:", imagen.shape)
```

```python
imagen[0:s,   0:s]   = [255, 0,   0  ]  
imagen[0:s,   s:2*s] = [255, 165, 0  ]  
imagen[0:s,   2*s:N] = [255, 255, 0  ]  

imagen[s:2*s, 0:s]   = [0,   255, 0  ]  
imagen[s:2*s, s:2*s] = [255, 255, 255]  
imagen[s:2*s, 2*s:N] = [0,   0,   255]  

imagen[2*s:N, 0:s]   = [0,   255, 255]  
imagen[2*s:N, s:2*s] = [128, 0,   128]  
imagen[2*s:N, 2*s:N] = [255, 105, 180]  

print("Imagen rellenada correctamente")
```

```python
plt.figure(figsize=(4, 4))
plt.imshow(imagen)
plt.title("Cara Cubo Rubik – 9 cuadrantes")
plt.axis("off")
plt.tight_layout()
plt.savefig("rubik.png", dpi=100)
plt.show()
```
