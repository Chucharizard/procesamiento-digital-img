# Ejercicio 1 - Factorial de un número entero

```python
def factorial(n):
    if n < 0:
        return None  
    resultado = 1
    for i in range(2, n + 1):
        resultado *= i
    return resultado

numero = int(input("Ingresa un número entero para calcular su factorial: "))

if numero < 0:
    print("El factorial no está definido para números negativos.")
else:
    print(f"El factorial de {numero} es: {factorial(numero)}")
```

# Ejercicio 2 - Número primo

```python
def es_primo(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


numero = int(input("Ingresa un número entero: "))

if es_primo(numero):
    print(f"{numero} ES un número primo.")
else:
    print(f"{numero} NO es un número primo.")
```

# Ejercicio 3 - Mayor y menor de N números

```python
limite = int(input("¿Cuántos números deseas ingresar? (mínimo 3): "))
while limite < 3:
    print("El límite debe ser al menos 3.")
    limite = int(input("¿Cuántos números deseas ingresar? (mínimo 3): "))

numeros = []
for i in range(1, limite + 1):
    num = float(input(f"Ingresa el número {i}: "))
    numeros.append(num)

print(f"\nNúmeros ingresados: {numeros}")
print(f"El número MAYOR es: {max(numeros)}")
print(f"El número MENOR es: {min(numeros)}")
```

# Ejercicio 4 - Palíndromo

```python
cadena = input("Ingresa una cadena de texto: ")

cadena_limpia = cadena.replace(" ", "").lower()

if cadena_limpia == cadena_limpia[::-1]:
    print(f'"{cadena}" ES un palíndromo.')
else:
    print(f'"{cadena}" NO es un palíndromo.')
```

# Ejercicio 5 - Número que más se repite

```python
numeros = []
for i in range(1, 8):
    num = int(input(f"Ingresa el número {i}: "))
    numeros.append(num)

conteo = {}
for num in numeros:
    conteo[num] = conteo.get(num, 0) + 1

mas_repetido = max(conteo, key=conteo.get)
veces = conteo[mas_repetido]

print(f"\nNúmeros ingresados: {numeros}")
print(f"El número que más se repite es: {mas_repetido} ({veces} veces)")
```

# Ejercicio 6 - Anagrama

```python
palabra1 = input("Ingresa la primera palabra: ").replace(" ", "").lower()
palabra2 = input("Ingresa la segunda palabra: ").replace(" ", "").lower()

if sorted(palabra1) == sorted(palabra2):
    print(f'"{palabra1}" y "{palabra2}" SÍ son anagramas.')
else:
    print(f'"{palabra1}" y "{palabra2}" NO son anagramas.')
```

# Ejercicio 7 - Ordenamiento burbuja

```python
def burbuja(lista):
    n = len(lista)
    for i in range(n - 1):
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
    return lista

cantidad = int(input("¿Cuántos números deseas ordenar? "))
numeros = []
for i in range(1, cantidad + 1):
    num = float(input(f"Ingresa el número {i}: "))
    numeros.append(num)

ordenados = burbuja(numeros.copy())
print(f"\nLista original: {numeros}")
print(f"Lista ordenada: {ordenados}")
```
