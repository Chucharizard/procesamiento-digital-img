```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def prosMost(ruta_imagen, nombre_guardado):
    # 1 cargamos la img en bgr 
    img_bgr = cv2.imread(ruta_imagen)

    # 2 convertimos a RGB 
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

    # 3 convertir a esc de grises
    img_gray = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2GRAY)

    # 4 guardar img en la pc (BGR y Grises)
    cv2.imwrite(f"{nombre_guardado}_bgr.jpg", img_bgr)
    cv2.imwrite(f"{nombre_guardado}_gris.jpg", img_gray)

    # 5 separon canales rgb
    b = img_bgr[:,:,0]
    g = img_bgr[:,:,1]
    r = img_bgr[:,:,2]
    
    plt.figure(figsize=(15, 10))

    #plot
    # 1 sin las comas "231"
    # org en formato de color
    plt.subplot(231) 
    plt.title('1. Original en Color (RGB)')
    plt.imshow(img_rgb)
    plt.axis('off')

    # 2 con las comas "2, 3, 2"
    # imagen en esc de grises pura
    plt.subplot(2, 3, 2) 
    plt.title('2. Escala de Grises')
    plt.imshow(img_gray, cmap='gray')
    plt.axis('off')

    # color falso 
    plt.subplot(2, 3, 3)
    plt.title('3. Falso Color (jet)')
    plt.imshow(img_gray, cmap='jet')
    plt.axis('off')

    # canal sep en escala de grises
    plt.subplot(2, 3, 4)
    plt.title('4. Canal Rojo')
    plt.imshow(r, cmap='gray')
    plt.axis('off')

    plt.subplot(2, 3, 5)
    plt.title('5. Canal Verde')
    plt.imshow(g, cmap='gray')
    plt.axis('off')

    plt.subplot(2, 3, 6)
    plt.title('6. Canal Azul')
    plt.imshow(b, cmap='gray')
    plt.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
# PRIMERA imagen
prosMost('foto1.jpg', 'mi_foto_1')
```

```python
#SEGUNDA imagen
prosMost('foto2.jpg', 'mi_foto_2')
```

```python
#TERCERA imagen
prosMost('foto3.jpg', 'mi_foto_3')
```
