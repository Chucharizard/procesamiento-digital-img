```python
import cv2
import numpy as np

# Lista de paletas de colores de OpenCV y sus nombres
colormaps = [
    (cv2.COLORMAP_AUTUMN,  "AUTUMN"),
    (cv2.COLORMAP_BONE,    "BONE"),
    (cv2.COLORMAP_JET,     "JET"),
    (cv2.COLORMAP_WINTER,  "WINTER"),
    (cv2.COLORMAP_RAINBOW, "RAINBOW"),
    (cv2.COLORMAP_OCEAN,   "OCEAN"),
    (cv2.COLORMAP_SUMMER,  "SUMMER"),
    (cv2.COLORMAP_SPRING,  "SPRING"),
    (cv2.COLORMAP_COOL,    "COOL"),
    (cv2.COLORMAP_HSV,     "HSV"),
    (cv2.COLORMAP_PINK,    "PINK"),
    (cv2.COLORMAP_HOT,     "HOT"),
]

# Crear una imagen en escala de grises con un degradado horizontal
alto = 200
ancho = 500
gradiente = np.linspace(0, 255, ancho, dtype=np.uint8)
img_gray = np.tile(gradiente, (alto, 1))  # (alto, ancho)

for cmap, nombre in colormaps:
    # Aplicar la paleta de colores
    img_color = cv2.applyColorMap(img_gray, cmap)

    # Mostrar la imagen en una ventana con el nombre de la paleta
    cv2.imshow(f"Colormap: {nombre}", img_color)

    # Esperar 1000 ms (1 segundo) o hasta que se presione una tecla
    key = cv2.waitKey(1000)

    # Cerrar la ventana de esta paleta
    cv2.destroyWindow(f"Colormap: {nombre}")

    # Si se presiona ESC, salir antes de terminar la lista
    if key == 27:  # 27 = tecla ESC
        break

cv2.destroyAllWindows()
```
