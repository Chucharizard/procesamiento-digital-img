```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucion_media_h(imagen, tkernel):
    kernel = np.ones((1, tkernel), np.float32) / tkernel
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

imagen = cv2.imread('img.jpg', 0)

img_fin = convolucion_media_h(imagen, 25)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.imshow(imagen, cmap='gray')
ax1.set_title('imagen original')
ax1.axis('off')

ax2.imshow(img_fin, cmap='gray')
ax2.set_title('convolucion media horizontal (k=7)')
ax2.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Version a colores
imagen_color = cv2.imread('img.jpg')
imagen_color = cv2.cvtColor(imagen_color, cv2.COLOR_BGR2RGB)

img_fin_color = convolucion_media_h(imagen_color, 7)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.imshow(imagen_color)
ax1.set_title('imagen original (color)')
ax1.axis('off')

ax2.imshow(img_fin_color)
ax2.set_title('convolucion media horizontal (k=7)')
ax2.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

random_img = np.random.randint(0, 256, size=(256, 256), dtype=np.uint8)
img_gaus = cv2.GaussianBlur(random_img, (9, 9), 0)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))

ax1.imshow(random_img, cmap='gray')
ax1.set_title('imagen aleatoria')
ax1.axis('off')

ax2.imshow(img_gaus, cmap='gray')
ax2.set_title('gaussian blur (9x9)')
ax2.axis('off')

plt.tight_layout()
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt 

def convolucion_media_h(imagen, tkernel):
    kernel = np.ones((1, tkernel), np.float32) / tkernel
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

def convolucion_media_v(imagen, tkernel):
    kernel = np.ones((tkernel, 1), np.float32) / tkernel
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

def convolucion_media_diag(imagen, tkernel):
    kernel = np.zeros((tkernel, tkernel), np.float32)
    for i in range(tkernel):
        kernel[i, i] = 1.0
    kernel = kernel / tkernel
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

def convolucion_media_diag_inversa(imagen, tkernel):
    kernel = np.zeros((tkernel, tkernel), np.float32)
    for i in range(tkernel):
        kernel[i, tkernel - 1 - i] = 1.0
    kernel = kernel / tkernel
    imagen_c = cv2.filter2D(imagen, -1, kernel)
    return imagen_c

img = cv2.imread('img.jpg')

img_h = convolucion_media_h(img, 30)
img_v = convolucion_media_v(img, 30)
img_d = convolucion_media_diag(img, 30)
img_di = convolucion_media_diag_inversa(img, 30)
plt.figure(figsize=(10, 16))
plt.subplot(4, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('imagen original')
plt.subplot(4, 2, 2)
plt.imshow(cv2.cvtColor(img_h, cv2.COLOR_BGR2RGB))
plt.title('imagen horizontal')

plt.subplot(4, 2, 3)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('imagen original')
plt.subplot(4, 2, 4)
plt.imshow(cv2.cvtColor(img_v, cv2.COLOR_BGR2RGB))
plt.title('imagen vertical')

plt.subplot(4, 2, 5)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('imagen original')
plt.subplot(4, 2, 6)
plt.imshow(cv2.cvtColor(img_d, cv2.COLOR_BGR2RGB))
plt.title('imagen diagonal')

plt.subplot(4, 2, 7)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title('imagen original')
plt.subplot(4, 2, 8)
plt.imshow(cv2.cvtColor(img_di, cv2.COLOR_BGR2RGB))
plt.title('imagen diagonal inversa')
plt.show()
```
